```python
import glob
import pandas as pd
from xml.dom import minidom

# Lấy file XML đầu tiên trong thư mục hiện tại
xml_file = glob.glob("*.xml")[0]

# Format XML
with open(xml_file, "r", encoding="utf-8") as f:
    xml = f.read()

pretty_xml = minidom.parseString(xml).toprettyxml(indent="    ")

# Tách từng dòng
rows = [x for x in pretty_xml.splitlines() if x.strip()]

# Xuất Excel
pd.DataFrame(rows, columns=["XML"]).to_excel(
    "xml_content.xlsx",
    index=False
)

print("Done!")
```
