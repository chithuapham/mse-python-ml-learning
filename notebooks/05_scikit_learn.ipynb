```python

from sklearn.linear_model import LinearRegression
import numpy as np

# Dữ liệu đầu vào
X = np.array([[1],[2],[3],[4]])

# Kết quả thực tế
y = np.array([100,120,140,160])

# Tạo model
model = LinearRegression()

# Huấn luyện model
model.fit(X, y)

# Dự đoán tháng 5
prediction = model.predict([[5]])

print(prediction)
```
