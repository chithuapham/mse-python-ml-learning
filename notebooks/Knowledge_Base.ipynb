```python
# --- 1. CƠ SỞ TRI THỨC (Knowledge Base) ---
RULES = [
    # Luật suy ra rủi ro
    {'if': ['yêu cầu không rõ', 'thời hạn gấp'],
     'then': 'RỦI RO: Cao'},

    {'if': ['nhóm thiếu kinh nghiệm', 'yêu cầu không rõ'],
     'then': 'RỦI RO: Rất cao'},

    {'if': ['ngân sách thấp'],
     'then': 'RỦI RO: Trung bình'},

    # Luật suy ra hành động (phản ứng dây chuyền)
    {'if': ['RỦI RO: Cao'],
     'then': 'HÀNH ĐỘNG: Cần họp khẩn với khách hàng'},

    {'if': ['RỦI RO: Rất cao'],
     'then': 'HÀNH ĐỘNG: Cảnh báo ban quản lý'},

    {'if': ['RỦI RO: Trung bình', 'thời hạn gấp'],
     'then': 'HÀNH ĐỘNG: Phải cắt giảm tính năng'}
]

# Nơi lưu trữ TẤT CẢ các sự thật đã biết (ban đầu rỗng)
KNOWN_FACTS = set()

# --- 2. BỘ MÁY SUY DIỄN (Inference Engine) ---

def run_inference_engine(new_fact):
    """
    Nhận MỘT sự thật mới, chạy Forward Chaining,
    và trả về danh sách các HỆ QUẢ MỚI vừa suy ra được.
    """

    # 1. Thêm sự thật mới vào cơ sở tri thức
    if new_fact not in KNOWN_FACTS:
        KNOWN_FACTS.add(new_fact)
        print(f"  [AI đang ghi nhận: '{new_fact}']")
    else:
        # Nếu gõ lại sự thật đã biết, không cần chạy lại
        return []

    newly_derived_facts = [] # Nơi lưu các kết luận MỚI để "trả lời"

    # 2. Bắt đầu vòng lặp suy diễn (hiệu ứng domino)
    facts_added_in_this_cycle = True
    while facts_added_in_this_cycle:
        facts_added_in_this_cycle = False

        for rule in RULES:
            conditions = rule['if']
            conclusion = rule['then']

            # 3. Nếu kết luận đã biết -> bỏ qua luật này
            if conclusion in KNOWN_FACTS:
                continue

            # 4. Kiểm tra xem tất cả điều kiện (vế IF) đã được đáp ứng chưa
            all_conditions_met = True
            for condition in conditions:
                if condition not in KNOWN_FACTS:
                    all_conditions_met = False
                    break # Chỉ cần 1 điều kiện sai -> bỏ qua

            # 5. KÍCH HOẠT LUẬT (nếu tất cả điều kiện đúng)
            if all_conditions_met:
                print(f"  [AI đang suy diễn: {conditions} => {conclusion}]")
                KNOWN_FACTS.add(conclusion)       # Thêm vào tri thức tổng
                newly_derived_facts.append(conclusion) # Thêm vào để "trả lời"
                facts_added_in_this_cycle = True       # Báo hiệu để lặp lại

    # 6. Trả về danh sách các kết luận mới
    return newly_derived_facts

# --- 3. GIAO DIỆN "CHAT" ---

def start_chat():
    print("=== CHÀO MỪNG BẠN ĐẾN VỚI AI TƯ VẤN RỦI RO DỰ ÁN ===")
    print("Hãy gõ vào các thông tin (sự thật) về dự án của bạn.")
    print("Ví dụ: 'thời hạn gấp', 'yêu cầu không rõ', 'nhóm thiếu kinh nghiệm'...")
    print("Gõ 'thoat' để kết thúc.")
    print("-" * 50)

    while True:
        # 1. Người dùng "gõ vào"
        user_input = input("Bạn (nhập sự thật): ").lower().strip()

        if user_input == 'thoat':
            print("AI: Tạm biệt. Chúc dự án thành công!")
            break

        if not user_input:
            continue

        # 2. AI "cho ra câu trả lời"
        conclusions = run_inference_engine(user_input)

        # 3. In ra kết quả
        if not conclusions:
            print("AI: Đã ghi nhận. (Tôi chưa suy ra được gì thêm từ thông tin này).")
        else:
            print("\nAI (vừa suy ra):")
            for i, con in enumerate(conclusions):
                print(f"  {i+1}. {con}")

        print("-" * 50)

# Bắt đầu chạy AI
start_chat()
```
