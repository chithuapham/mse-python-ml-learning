```python
# Bước 1: Import các thư viện cần thiết
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# Bước 2: Tạo dữ liệu mẫu
data = {
    'Dien_tich_m2': [50, 70, 100, 120, 150, 200, 80, 90, 110, 130],
    'So_phong_ngu': [1, 2, 3, 3, 4, 5, 2, 2, 3, 4],
    'Khoang_cach_km': [5, 3, 2, 1, 4, 6, 3, 2, 5, 1],
    'Gia_nha_trieu': [150, 200, 300, 350, 400, 500, 220, 250, 320, 380]
}
df = pd.DataFrame(data)

# Bước 3: Tách đặc trưng và nhãn
X = df[['Dien_tich_m2', 'So_phong_ngu', 'Khoang_cach_km']]  # Đặc trưng
y = df['Gia_nha_trieu']  # Nhãn

# Bước 4: Chia dữ liệu thành tập train và test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Bước 5: Khởi tạo và huấn luyện mô hình
model = LinearRegression()
model.fit(X_train, y_train)

# Bước 6: Dự đoán và đánh giá mô hình
y_pred = model.predict(X_test)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R² Score: {r2:.2f}")

# Bước 7: Hiển thị hệ số hồi quy
print("\nHệ số hồi quy:")
for feature, coef in zip(X.columns, model.coef_):
    print(f"{feature}: {coef:.2f}")
print(f"Hằng số (intercept): {model.intercept_:.2f}")

# Bước 8: Dự đoán giá nhà mới
new_house = np.array([[60, 2, 12]])  # Diện tích 100m², 3 phòng, cách 2km
new_house_df = pd.DataFrame(new_house, columns=X.columns)  # Chuyển thành DataFrame
predicted_price = model.predict(new_house_df)
print(f"\nGiá nhà dự đoán: {predicted_price[0]:.2f} triệu")
```
